## Data Wrangling — Library Setup

In this section, the required libraries for data manipulation and preprocessing are loaded. These libraries provide functions for data cleaning, transformation, and preparation.

- `dplyr` is used for data manipulation
- `tidyr` is used for reshaping data
- `lubridate` is used for handling date-time values

```r
# Load required libraries for data wrangling
library(dplyr)        # Used for data manipulation operations such as filtering and grouping
library(tidyr)        # Used for handling missing values and reshaping data
library(lubridate)    # Used for working with date and time features
```

### Load Raw Dataset

The raw Seoul bike-sharing dataset is loaded from the data directory. This dataset contains historical records of bike rentals along with weather and temporal variables.

```r
# Load raw bike-sharing dataset from repository path
bike_data <- read.csv("seoul_bike_sharing.csv")  # Read CSV file into R dataframe

# Display first few rows to inspect structure
head(bike_data)  # View top rows of dataset

# Check structure of dataset
str(bike_data)  # Display column types and structure
```

## Handling Missing Values

Missing values can negatively impact analysis and modeling. In this step, we identify and remove or handle missing values to ensure data consistency and reliability.

```r
# Check for missing values in each column
colSums(is.na(bike_data))  # Count number of NA values per column

# Remove rows with missing values
bike_data <- na.omit(bike_data)  # Drop rows containing any NA values
```

## Standardizing Column Names

Column names are standardized to lowercase with underscores for consistency and compatibility with SQL queries.

```r
# Convert column names to lowercase and replace spaces with underscores
colnames(bike_data) <- tolower(gsub(" ", "_", colnames(bike_data)))  # Clean column names
```

## Data Type Conversion

To ensure proper analysis, relevant columns are converted to appropriate data types such as numeric and date formats.

```r
# Check missing values per column
colSums(is.na(bike_data))

# Remove rows with missing values
bike_data_clean <- na.omit(bike_data)

# Verify cleaned dataset
colSums(is.na(bike_data_clean))
```

## Convert Data Types

Some columns may not be in the correct format.

We ensure:

- Numerical variables are numeric
- Categorical variables are factors

```r
# Convert date column to Date format
bike_data$date <- as.Date(bike_data$date, format = "%d/%m/%Y")  # Convert string to Date type

# Ensure numeric columns are properly typed
bike_data$rented_bike_count <- as.numeric(bike_data$rented_bike_count)  # Convert rentals to numeric
bike_data$temperature <- as.numeric(bike_data$temperature)              # Convert temperature to numeric
bike_data$humidity <- as.numeric(bike_data$humidity)                    # Convert humidity to numeric
bike_data$wind_speed <- as.numeric(bike_data$wind_speed)                # Convert wind speed to numeric
```

## Feature Engineering

New features are derived to support analysis and modeling. These include temporal features such as year, month, and day extracted from the date column.

```r
# Extract year from date
bike_data$year <- year(bike_data$date)  # Create year feature

# Extract month from date
bike_data$month <- month(bike_data$date)  # Create month feature

# Extract day of week
bike_data$day_of_week <- weekdays(bike_data$date)  # Create categorical weekday feature
```

## Final Clean Dataset

After preprocessing, the cleaned dataset is saved for downstream analysis including SQL-based EDA and modeling.

```r
# Save cleaned dataset to processed folder
write.csv(bike_data, "clean_bike_data.csv", row.names = FALSE)  # Export clean dataset

# Confirm dataset saved successfully
file.exists("clean_bike_data.csv")  # Check file existence
```

In this notebook, the raw Seoul bike-sharing dataset was systematically prepared for downstream analysis and modeling.

The data wrangling process included cleaning the dataset by handling missing values and removing inconsistencies, followed by standardizing column names to ensure compatibility with SQL-based analysis. Regular expressions were applied where necessary to refine and format data fields.

Feature engineering techniques were used to derive additional variables, and categorical variables were transformed into dummy variables to support modeling requirements. Numerical features were also normalized to ensure consistent scaling across variables.
