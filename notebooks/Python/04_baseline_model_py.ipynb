# Baseline Regression Model (Weather Variables Only)

After completing exploratory data analysis, the next step is to build an initial predictive model. This baseline model serves as a reference point for evaluating future improvements.

The goal of this notebook is to predict hourly bike-sharing demand using only weather-related variables. This allows us to isolate the effect of environmental conditions before introducing more complex features such as time and seasonal variables.

According to the lab structure, the modeling workflow includes:

- Splitting the dataset into training and testing subsets
- Building a linear regression model using weather variables
- Evaluating model performance using standard metrics

This baseline model provides a foundation for iterative improvements in later stages.

## Import Required Libraries

We use the following libraries for modeling:

- `pandas`: data handling
- `numpy`: numerical computations
- `scikit-learn`: machine learning modeling and evaluation

```python
# Import libraries for modeling

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split    # For splitting dataset
from sklearn.linear_model import LinearRegression       # Linear regression model
from sklearn.metrics import mean_squared_error, r2_score  # Evaluation metrics
```

## Load Dataset and Sort Chronologically

We load the cleaned (pre-normalization) dataset produced by notebook 02 and immediately sort by date. Sorting up front means the chronological train/test split below is just a positional slice, not a time-aware reshuffle. The CSV stores date strings as `dd/mm/yyyy`, so we parse with `dayfirst=True` to avoid silently flipping day and month.

```python
# ── Load and Sort Cleaned Dataset ─────────────────────────────
df = pd.read_csv("cleaned_bike_sharing.csv")                  # load wrangled snapshot from notebook 02
df.head()                                                     # peek at first rows to confirm load
df['date'] = pd.to_datetime(df['date'], dayfirst=True)        # parse date strings as dd/mm/yyyy (Korean source format)
df = df.sort_values('date').reset_index(drop=True)            # sort chronologically and reset index for positional slicing
```

## Select Features and Target Variable

In this baseline model, we only use weather-related variables as predictors.

The target variable is:
- `rented_bike_count`

Selected features include:
- Temperature
- Humidity
- Wind speed
- Visibility

```python
# Define features and target

X = df[['temperature', 'humidity', 'wind_speed', 'visibility']]
y = df['rented_bike_count']
```

## Chronological Train/Test Split

We split chronologically (earliest 80% → train, latest 20% → test) rather than randomly. A time-aware split is the honest evaluation for a forecasting workflow — it prevents the baseline from peeking at future demand patterns. Because we already sorted by `date`, the split is just a positional slice.

```python
# ── Chronological 80/20 Split ─────────────────────────────────
split_idx = int(len(df) * 0.8)                                # index of last training row (first 80%)
X_train, X_test = X.iloc[:split_idx], X.iloc[split_idx:]      # earliest 80% of features → train, latest 20% → test
y_train, y_test = y.iloc[:split_idx], y.iloc[split_idx:]      # mirror the same positional slice for the target
```

## Train Linear Regression Model

We fit a linear regression model using the training dataset. This model attempts to learn a linear relationship between the weather variables and bike demand.

```python
# Initialize and train model

model = LinearRegression()

model.fit(X_train, y_train)
```

## Model Coefficients

The learned coefficients indicate how each feature influences the target variable.

Positive coefficients indicate a direct relationship, while negative coefficients indicate an inverse relationship.

```python
# Display coefficients

coefficients = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': model.coef_
})

coefficients
```

## Make Predictions

We generate predictions using the test dataset to evaluate how well the model performs on unseen data.

```python
# Make predictions

y_pred = model.predict(X_test)
```

## Model Evaluation

We evaluate model performance using:

- R-squared (R²): Measures proportion of variance explained
- RMSE: Measures prediction error magnitude

```python
# Calculate evaluation metrics

r2 = r2_score(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print("R-squared:", r2)
print("RMSE:", rmse)
```

## Interpretation

The baseline model provides a reference for evaluating improvements in future models.

Key observations:

- Weather variables explain a portion of bike demand variability
- Model performance is limited due to missing temporal and behavioral factors
- Additional variables are required to improve predictive accuracy

This motivates the development of more advanced models in subsequent notebooks.
